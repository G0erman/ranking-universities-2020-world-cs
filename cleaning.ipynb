```python
with open('data/datalake/bronze/timeshighereducation_CS_2022_original') as f:
    lines = f.readlines()
```

Logic

```python
u = lines[0] + lines[1] + lines[2]
u
```

```python
final_u = u.replace('\n','\t', 2)
final_u
```

```python
print(final_u)
```

Procedure

```python
parsed_universities = []
for i, line in enumerate(lines, 1):
    if i % 3 == 0:
        u = lines[i-3] + lines[i-2] + lines[i-1]
        final_u = u.replace('\n','\t', 2)
        parsed_universities.append(final_u)
        print(i, final_u)
```

```python
parsed_universities
```

```python
title = 'Rank	Name	Country/Region	No. of FTE Students	No. of students per staff	International Students	Female:Male Ratio\n'
```

```python
parsed_universities.insert(0, title)
```

```python
with open('data/datalake/silver/timeshighereducation_CS_2022.tsv','w') as f:
    f.writelines(parsed_universities)
```
